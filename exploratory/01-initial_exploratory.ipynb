## mini experience: follow the bus

```python
import time

bus = "1298"

# check the bus every 20 seconds, for 5 minutes
for i in range(15):
    v = pd.DataFrame(requests.get(f"{api}/vehicles").json())
    row = v[v["id"].str.contains(bus)].iloc[0]
    now = datetime.now().strftime("%H:%M:%S")
    print(now, row["stop_id"], row["current_status"], row["speed"])
    time.sleep(20)
```

```python
# call every 5 seconds for 2 minutes and see how often the positions really change
last = None

for i in range(24):
    v = pd.DataFrame(requests.get(f"{api}/vehicles").json())
    newest = v["timestamp"].max()
    now = datetime.now().strftime("%H:%M:%S")

    changed = "new data" if newest != last else "same as before"
    print(now, newest, changed)

    last = newest
    time.sleep(5)
```

# summary of each available dataset

```python
def summary(df):
    # one row per column: type, % empty, how many different values, one example
    return pd.DataFrame({
        "type": df.dtypes.astype(str),
        "empty_%": (df.isna().mean() * 100).round(1),
        "unique": df.astype(str).nunique(),
        "example": df.apply(lambda c: c.dropna().iloc[0] if c.notna().any() else None),
    })
```

```python
datasets = {
    "arrivals": df,
    "stops": stops,
    "vehicles": vehicles,
    "districts": districts,
    "municipalities": municipalities,
}

for name, data in datasets.items():
    print(f"{name} {data.shape}")
    print(summary(data))
```

**vehicles**

* **keep**: id, agency_id, line_id, route_id, pattern_id, trip_id, direction_id,
      stop_id, current_status, timestamp, speed, bearing, lat, lon
* **drop** (always empty): block_id, shift_id, emission_class, license_plate, make, model,
      owner, propulsion, registration_date, wheelchair_accessible, occupancy_estimated
* **drop** (same value for every bus): capacity_*, bikes_allowed, contactless, door_status,
      occupancy_status
* **not sure**: schedule_relationship (always SCHEDULED now, but might change when a trip is cancelled),
      event_id (unique per row, could help remove duplicates)

**stops**
