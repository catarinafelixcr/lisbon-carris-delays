# Here we must undersand the data that we have!

```python
import pandas as pd
import requests

api = "https://api.carrismetropolitana.pt/v2"
```

```python
# oeiras train station, lots of lines stop here
stop_id = "121270"

r = requests.get(f"{api}/arrivals/by_stop/{stop_id}")
df = pd.DataFrame(r.json())
```

```python
df.head(5)
```

```python
df.info()
```

```python
df.shape  # this means that we have a lot of features without information at all
```

```python
# how many empty values per column
df.isna().sum()
```

```python
# how many different values per column
df.nunique()
```

```python
# the 4 rows that have live info, next to the time right now
from datetime import datetime
print(datetime.now())

live = df[df["estimated_arrival"].notna()]
live[["line_id", "scheduled_arrival", "estimated_arrival", "trip_id", "vehicle_id"]]
```

## lets see **districts** and **municipalities** information

```python
# district codes -> names
districts = pd.DataFrame(requests.get(f"{api}/locations/districts").json()["data"])
districts
```

```python
districts.shape
```

we have 29 districts!! so its not only about lisbon!!

```python
# municipality codes -> names
municipalities = pd.DataFrame(requests.get(f"{api}/locations/municipalities").json()["data"])
municipalities.head(20)
```

```python
municipalities.shape
```

```python
district_lisbon = municipalities['district_id'] == '11'  # lisbon
municipalities[district_lisbon]
```
